```python
import numpy as np
from tensorflow.keras.datasets import imdb

num_words = 10000
(train_data, train_labels), (test_data, test_labels) = imdb.load_data(num_words=num_words)

vocab = range(num_words)

input_dataset = [list(set(review)) for review in train_data]
target_dataset = list(train_labels)

test_input_dataset = [list(set(review)) for review in test_data]
test_target_dataset = list(test_labels)

np.random.seed(42)

def sigmoid(x):
  return 1 / (1 + np.exp(-x))

alpha, iterations = (0.01, 100)
hidden_size = 100

weights_0_1 = (np.random.random((len(vocab), hidden_size)) - 0.5) * 0.2
weights_1_2 = (np.random.random((hidden_size, 1)) - 0.5) * 0.2
for iter in range(iterations):
  correct = 0
  # Training loop
  for i in range(len(input_dataset)):
    x, y = (input_dataset[i], target_dataset[i])
    layer_1 = sigmoid(np.sum(weights_0_1[x], axis=0))
    layer_2 = sigmoid(np.dot(layer_1, weights_1_2))
    layer_2_delta = layer_2 - y
    layer_1_delta = layer_2_delta.dot(weights_1_2.T)
    weights_0_1[x] -= layer_1_delta * alpha
    layer_1 = layer_1.reshape(hidden_size, 1)
    layer_2_delta = layer_2_delta.reshape(1, 1)
    weights_1_2 -= np.dot(layer_1, layer_2_delta) * alpha
    if (np.abs(layer_2_delta) < 0.5):
      correct += 1
  print("Iteration " + str(iter + 1) + " Training Accuracy: " + str(correct / len(input_dataset)))

#Testing Phase
correct, total = (0, 0)
for i in range(len(test_input_dataset)):
  x = test_input_dataset[i]
  y = test_target_dataset[i]
  layer_1 = sigmoid(np.sum(weights_0_1[x], axis=0))
  layer_2 = sigmoid(np.dot(layer_1, weights_1_2))
  if (np.abs(layer_2 - y) < 0.5):
    correct += 1
  total += 1
print("Testing Accuracy Accuracy: " + str(correct / total))
```
