```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_csv("winequality-total.csv", sep=';')

X_df = df.drop('quality', axis=1)
y_df = df['quality']

X_train, X_test, y_train, y_test = train_test_split(X_df, y_df, test_size=0.2, random_state=42)
```

```python
print("--- Running Linear Regression with Scikit-Learn ---")
scaler = StandardScaler()

X_train_scaled_sk = scaler.fit_transform(X_train)
X_test_scaled_sk = scaler.transform(X_test)

model_sk = LinearRegression()
model_sk.fit(X_train_scaled_sk, y_train)

y_pred_sk = model_sk.predict(X_test_scaled_sk)

intercept_sk = model_sk.intercept_

mse_sk = mean_squared_error(y_test, y_pred_sk)
r2_sk = r2_score(y_test, y_pred_sk)

print(f"Intercept: {intercept_sk:.4f}")
print(f"Mean Squared Error (MSE): {mse_sk:.3f}")
print(f"R-squared (R2): {r2_sk:.3f}")
```

```python
print("--- Running Linear Regression with NumPy ---")

class LinearRegression:
    def __init__(self, lr=0.01, np_iter=5000):
        self.lr = lr
        self.np_iter = np_iter
        self.weights = None

    def fit(self, X_train, y_train):
        y_train_vec = y_train.reshape(-1, 1)
        num_samples, num_features = X_train.shape
        self.weights = np.zeros((num_features, 1))

        for i in range(self.np_iter):
            y_pred = X_train @ self.weights
            error = y_pred - y_train_vec
            dw = (1 / num_samples) * (X_train.T @ error)
            self.weights = self.weights - self.lr * dw

        return self

    def predict(self, X):
        return X @ self.weights


X_train_np = X_train.values
X_test_np = X_test.values
y_train_np = y_train.values
y_test_np = y_test.values

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_np)
X_test_scaled = scaler.transform(X_test_np)

X_train_b = np.c_[np.ones((X_train_scaled.shape[0], 1)), X_train_scaled]
X_test_b = np.c_[np.ones((X_test_scaled.shape[0], 1)), X_test_scaled]

model_gd = LinearRegression(lr=0.01, np_iter=5000)
model_gd.fit(X_train_b, y_train_np)

y_pred_gd = model_gd.predict(X_test_b).flatten()

mse_gd = mean_squared_error(y_test_np, y_pred_gd)
r2_gd = r2_score(y_test_np, y_pred_gd)

intercept_gd = model_gd.weights[0][0]

print(f"Intercept: {intercept_gd:.4f}")
print(f"Mean Squared Error (MSE): {mse_gd:.3f}")
print(f"R-squared (R2): {r2_gd:.3f}")
```
