```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("winequality-total.csv", sep = ";")

y = df['quality'].values
X = df.drop('quality', axis = 1).values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
class PCAnumpy:
  def __init__(self, n_components = 1, mean = None):
    self.n_components = n_components
    self.mean = mean

  def fit_transform(self, X):
    self.mean = np.mean(X, axis = 0)
    X_centered = X - self.mean
    cov = np.corrcoef(X_centered.T)
    eigenvalues, eigenvectors = np.linalg.eig(cov)
    indices = np.argsort(eigenvalues)[::-1]
    sorted_eigenvalues = eigenvalues[indices]
    sorted_eigenvectors = eigenvectors[:, indices]

    new_data = np.dot(X_centered, sorted_eigenvectors)
    new_data = -new_data
    return new_data[:, :self.n_components]

model = PCAnumpy(n_components = 1)
y_pred = model.fit_transform(X_scaled)
print(y_pred)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components = 1)
y_predsk = pca.fit_transform(X_scaled)
print(y_predsk)
```
