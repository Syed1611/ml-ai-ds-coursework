```python
#Steps to PCA

#1.Centering the values aka
    #mean of both features
    #subtract the mean form the features to get the targets
#2. Calculating the covariance Matrix
    #calculating the eigen values
#4. Calculate the eigen vectors
#5. Calculate the principal components
```

```python
import numpy as np

X = np.array([[126, 78],
              [128, 80],
              [128, 82],
              [130, 82],
              [130, 84],
              [132, 86]])
```

```python
class PCAnumpy:
  def __init__(self, n_components = 1, mean = None):
    self.n_components = n_components
    self.mean = mean

  def fit_transform(self, X):
    self.mean = np.mean(X, axis = 0)
    X_centered = X - self.mean
    cov = np.cov(X_centered.T, ddof = 1)
    eigenvalues, eigenvectors = np.linalg.eig(cov)
    indices = np.argsort(eigenvalues)[::-1]
    sorted_eigenvalues = eigenvalues[indices]
    sorted_eigenvectors = eigenvectors[:, indices]
    for i in range(sorted_eigenvectors.shape[1]):
      max_abs_idx = np.argmax(np.abs(sorted_eigenvectors[:, i]))
      if sorted_eigenvectors[max_abs_idx, i] < 0:
        sorted_eigenvectors[:, i] *= -1

    new_data = np.dot(X_centered, sorted_eigenvectors)
    return new_data[:, :self.n_components]
```

```python
model = PCAnumpy()
y_pred = model.fit_transform(X)
print(y_pred)

```

```python
from sklearn.decomposition import PCA
pca = PCA()
y_pred = pca.fit_transform(X)
print(y_pred)
```
